```python
import asyncio
import ipywidgets as widgets
from IPython.display import display
```

```python
btn = widgets.Button(description="Continue", button_style="primary", disabled=True)
out = widgets.Output()
display(out, btn)


def wait_click():
    future = asyncio.Future()

    def on_click(b):
        btn.on_click(on_click, remove=True)
        future.set_result(True)

    btn.on_click(on_click)
    return future


async def _run(iterator):
    btn.disabled = False
    with out:
        print("Started...")
    for res in iterator:
        with out:
            print("res:", res)
        await wait_click()
    with out:
        print("...Done")
    btn.disabled = True


def run(iterator):
    return asyncio.create_task(_run(iterator))
```

```python
def iterator():
    yield 1
    yield 2
    yield 3
    yield 4
```

```python
run(iterator())
```
