```python
def iter_subblock(blk: int) -> Iterable[Locality]:
    r1, c1 = Loc.of_blk(blk)
    yield from (Locality(blk, r, None) for r in (r1, r1 + 1, r1 + 2))
    yield from (Locality(blk, None, c) for c in (c1, c1 + 1, c1 + 2))
```

```python
from collections import Counter
from analytics import MultiTarget
from board import POS9


def count_digits(nodes: Iterable[Node]):
    return Counter(iterflat(n.cell for n in nodes))


def search_groups(board: Board) -> Iterable[MultiTarget]:
    for blk in POS9:
        loc = Locality(blk, None, None)
        counts = count_digits(filter(Node.is_draft, loc.neighborhood(board)))
        print(counts)
        for key, cnt in counts.items():
            if cnt > 2:
                yield MultiTarget(loc, frozenset((key,)))
```
